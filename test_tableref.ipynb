### Candidate generation

We use an LLM to extract simple references (surname and data) for later mathing with references from the list of references of the paper under analysis

```python
from tableref.candidates import OllamaCandidateGenerator
from tableref.config import OllamaConfig


cgen = OllamaCandidateGenerator(OllamaConfig)
candidates_lststr = cgen.generate(res_img2table[0].get("df"))
candidates_lststr
```

```python
# Notice candidaets are returned as a list with one string. Has to be split before matching.
candidates = candidates_lststr[0].split(",")
candidates
```

### Matching

Then we, match the references extracted from the table to the ones in the reference section.


Matching of references tables (extracted from papers using the classes form the sections above) to references (lit extracted separately, e.g., using GROBID).

```python
from episcope.db.in_memory_academic_db import InMemoryAcademicDB
db = InMemoryAcademicDB("test.json")

paper_id = "2022_Du_k"
data_type="references"
strategy_name="grobid_test"

references = db.retrieve(paper_id, data_type, strategy_name)
[r.to_dict() for r in references]
```

```python
from tableref.matching import ReferenceMatcher
from tableref.config import SimpleSurnameMatcherConfig
matcher = ReferenceMatcher(config=SimpleSurnameMatcherConfig())
matcher.prepare(references)
```

```python
matcher.match_candidates(candidates, list(references), top_n=1)
```

### Full pipeline

```python
from tableref.pipeline import TableRefPipeline
pipeline = TableRefPipeline(
    extractors= [Img2TableExtractor()],
    candidate_generator= OllamaCandidateGenerator(OllamaConfig),
    matcher= ReferenceMatcher(SimpleSurnameMatcherConfig()),
    output_root="table_output/",
    )
results = pipeline.run(
    "pdfs/2022_Du_k.pdf", 
    references=list(references)
    )
results
```
